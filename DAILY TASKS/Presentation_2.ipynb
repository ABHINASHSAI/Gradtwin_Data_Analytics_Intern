**Collections :** A collection in Python is a data structure used to store and organize multiple elements or values in a single variable, allowing them to be accessed and manipulated efficiently.

**Type of  Collections (Data struture):**

Collection	Ordered	Changeable	Allows duplicates	Example

Tuple	        ✅	     ❌	          ✅	         (10, 20, 10)

**1.List** : A List in Python is an ordered and changeable collection of elements that can store multiple values, including duplicate values, in a single variable.

```python
course = ["Python", "SQL", "PowerBI"]

print(course)
```

1.1 list Initializing and adding a list

```python
course = ["Python", "SQL"]
print(course)

course.append("Tableau")
print(course)

course.extend(["Excel", "Statics"])
print(course)

course.insert(2, "Powerbi")
print(course)
```

1.2 List Removing and deleting

```python
course = ["Python", "SQL"]
print(course)

course.append("Tableau")
print(course)

course.extend(["Excel", "Statics"])
print(course)

course.insert(2, "Powerbi")
print(course)

print("Removing and Deleting in list")

course.pop()
print(course)

del course[0]
print(course)

course.remove("SQL")
print(course)

```

1.3 Sorting

```python
numbers = [17,1,7,0,7,3,8,23]

print(numbers)

numbers.reverse()
print(numbers)

numbers.sort()
print(numbers)

numbers.sort(reverse=True)
print(numbers)
```

1.4 Slicing

```python
numbers = [20,45,17,98,45,73,97,20,1]

print(numbers)
print(numbers[2:5])
print(numbers[:5])
print(numbers[5:])
```

**2.Tuple : **A Tuple in Python is an ordered and unchangeable collection of elements that can store multiple values, including duplicate values, in a single variable.

```python
numbers = (10, 20, 30)

print(numbers)
print(numbers[1])
```

2.1 InInitializing a Tuples


```python
numbers = (10, 20, 30)

print(numbers)
print(numbers[1])
```

2.2 Length and count of the tuple

```python
numbers = (10, 20, 30, 10, 30, 10, 50, 30)
print(numbers)
print(len(numbers))
print(numbers.count(10))
```

2.3 Adding 2 tuples into 1

```python
data_type1 = ("Int", "Float")
data_type2 = ("String","Complex")

data_type = data_type1 + data_type2
print(data_type)
```

2.4 Slicing in tuples

```python
months = ("Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec")

print(months)
print(months[2])
print(months[2:5])
print(months[:5])
print(months[5:])
```

**3.Set : **A Set in Python is an unordered and changeable collection of unique elements that does not allow duplicate values.

```python
numbers = {10,20,10,30}
print(numbers)
```

3.1 initialzing and length of the set

```python
n = {1, 2, 3, 4, 5, 2, 3}
print(n)
print("Length of the set:", len(n))
```

3.2 Adding and Removing in Set

```python
n = {1, 2, 3, 4, 5, 2, 3}
print(n)
n.add(6)

print(n)

n.remove(3)

print(n)

n.update([7, 8, 9])

print(n)
```

3.3 Slicing of set

```python
n = {7,17,9,6,4,3,8,2,9,7,1}
print(n)

print(n[2])
```

**4.Dictionary :** A Dictionary in Python is a changeable collection that stores data in the form of key-value pairs.

```python
employee = {
    "emp_name": "Ram",
    "emp_age": 23,
    "emp_id" : 1710,
    "emp_role" : "Data Analytics",
    "emp_salary" : 30000
}

print(employee)

employee['emp_salary'] = 35000

print("Updated :",employee)

employee ["emp_no"] = [9989900]

print(employee)

employee.pop ("emp_no")

print(employee)
```
