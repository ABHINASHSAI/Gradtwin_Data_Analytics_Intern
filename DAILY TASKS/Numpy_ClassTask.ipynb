**Numpy in Python**

```python
import numpy as np

numbers = np.array([10, 20, 30, 40])

print(numbers)

print(type(numbers))

print(numbers.ndim)

print(numbers.size)
```

1. Create arrays from lists

```python
import numpy as np

a = np.array([1, 2, 3])
print(a)
```

2. Create a 2D array

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
print(a)
```

3. Array of zeros

```python
import numpy as np

print(np.zeros(7))
```

4. Array of ones

```python
import numpy as np

print(np.ones((3, 2), dtype=int))
```

5. A range of values – arange()

```python
import numpy as np

print(np.arange(1, 15, 2))
```

6. Evenly spaced values – linspace()

```python
import numpy as np

print(np.linspace(0, 5, 5))
```

7. Identity matrix – eye()

```python
import numpy as np

print(np.eye(5, dtype=int))
```

Array Attributes

```python
import numpy as np

a = np.array([[10, 20, 30, 40],
              [50, 60, 70, 80]])

print("Shape:", a.shape)
print("Dimensions:", a.ndim)
print("Size:", a.size)
print("Data type:", a.dtype)
print("Rows:", a.shape[0])
print("Columns:", a.shape[1])
```

Convert Data Type – astype()

```python
import numpy as np

a = np.array([10, 20, 30, 40])

b = a.astype(float)

print(b)
print(b.dtype)
```

1D Array Indexing

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(a[0])
print(a[2])
print(a[4])
```

1D Array Slicing

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50, 60])

print(a[1:4])
print(a[0:3])
print(a[2:])
print(a[:4])
```

2D Array Indexing

```python
import numpy as np

a = np.array([[10, 20, 30],
              [40, 50, 60]])

print(a[0, 1])
print(a[1, 2])
```

2D Array Slicing

```python
import numpy as np

b = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

print(b[0:2, 0:2])
print(b[0:2, :])
print(b[:, 1])
print(b[1, :])
print(b[:, :])
```

**Arithmetic Operations**

```python
import numpy as np

a = np.array([50, 40, 30, 20])

print(a - 10)
```

```python
import numpy as np

a = np.array([2, 4, 6, 8])

print(a * 5)
```

4. Division

5. Power

```python
import numpy as np

a = np.array([2, 3, 4, 5])

print(a ** 2)
```
