# Pandas – Section A: Series and DataFrame


```
import pandas as pd
```

1. Create a Pandas Series using the values [10, 20, 30, 40, 50].

```
values = [10, 20, 30, 40, 50]

s = pd.Series(values)
print(s)
```

2. Create a DataFrame with three columns: Name, Age, and City using at least five student records.

```
data = {
    'Name': ['Ram', 'Raj', 'Sai', 'Hari', 'Bharath'],
    'Age': [20, 21, 22, 20, 21],
    'City': ['Chennai', 'Hyderabad', 'Vijayawada', 'Bengaluru', 'Peddapuram']
}

df = pd.DataFrame(data)
print(df)
```

3. Create a DataFrame using a dictionary containing student names, marks, and departments.

```
data = {
    'Names': ['Arun', 'Kiran', 'Ravi', 'Meena', 'Priya'],
    'Marks': [85, 90, 78, 92, 88],
    'Departments': ['CSE', 'AI', 'ECE', 'CSE', 'AI']
}

df = pd.DataFrame(data)
print(df)
```

4. Display the first 5 rows of a DataFrame.

```
print(df.head(5))
```

5. Display the last 3 rows of a DataFrame.

```
print(df.tail(3))
```

6. Display the number of rows and columns in a DataFrame.

```
print(df.shape)
```

7. Display all column names from a DataFrame.

```
print(df.columns)
```

8. Find the data types of all columns.

```
print(df.dtypes)
```

9. Display the basic statistical summary of numerical columns.

```
print(df.describe())
```

10. Check whether a DataFrame contains any missing values.

```
print(df.isnull().sum())
```

11. Select only the Name column from a DataFrame.

```
print(df['Names'])
```

12. Select the Name and Marks columns.

```
print(df[['Names', 'Marks']])
```

13. Display the records of students whose marks are greater than 80.

```
print(df[df['Marks'] > 80])
```

14. Filter students whose age is below 25.

```
print(df[df['Marks'] < 90])
```

15. Display students who belong to the CSE department.

```
print(df[df['Departments'] == 'CSE'])
```
