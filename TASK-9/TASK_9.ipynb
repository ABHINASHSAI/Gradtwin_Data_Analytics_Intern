# SECTION A – DATAFRAME BASICS

### Q1. Create a DataFrame using the following employee details.

```python
import pandas as pd

data_a = {
    "Name": ["Arun", "Priya", "Karthik", "Divya", "Rahul"],
    "Age": [23, 25, 22, 24, 26],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [35000, 30000, 28000, 40000, 32000]
}

df_a = pd.DataFrame(data_a)
print(df_a)
```

### Q2. Display the first 3 rows and the last 2 rows.

```python
print("First 3 rows:")
print(df_a.head(3))

print("Last 2 rows:")
print(df_a.tail(2))
```

### Q3. Display the number of rows and columns, column names, data types and statistical summary.

```python
print("Number of rows and columns:")
print(df_a.shape)

print("Column names:")
print(df_a.columns)

print("Data types:")
print(df_a.dtypes)

print("Statistical summary:")
print(df_a.describe())
```

### Q4. Display only the Name and Salary columns.

```python
print(df_a[["Name", "Salary"]])
```

### Q5. Display employees whose salary is greater than 30,000.

```python
print(df_a[df_a["Salary"] > 30000])
```

# SECTION B – DATA SELECTION AND FILTERING

### Q6. Display employees whose salary is greater than 40,000.

```python
data_b = {
    "Employee": ["Arun", "Bala", "Charan", "Divya", "Ezhil", "Farah"],
    "Age": [25, 30, 28, 24, 32, 27],
    "Salary": [30000, 50000, 45000, 25000, 60000, 40000],
    "Experience": [2, 5, 4, 1, 7, 3]
}

df_b = pd.DataFrame(data_b)
print(df_b[df_b["Salary"] > 40000])
```

### Q7. Filter employees whose age is between 25 and 30, including both limits.

```python
print(df_b[df_b["Age"].between(25, 30)])
```

### Q8. Find employees whose salary is above 30,000 and experience is greater than 3 years.

```python
print(df_b[(df_b["Salary"] > 30000) & (df_b["Experience"] > 3)])
```

### Q9. Display the first 3 records using iloc and retrieve Divya's record using loc.

```python
print("First 3 records using iloc:")
print(df_b.iloc[:3])

print("Divya's record using loc:")
print(df_b.loc[df_b["Employee"] == "Divya"])
```

# SECTION C – DATA CLEANING AND TRANSFORMATION

### Q10. Display the missing-value count for every column.

```python
data_c = {
    "Name": ["Arun", "Priya", "Karthik", "Divya", "Rahul"],
    "Age": [25, float("nan"), 28, 24, float("nan")],
    "Salary": [30000, 35000, float("nan"), 40000, 32000],
    "Department": ["IT", "HR", "IT", float("nan"), "HR"]
}

df_c = pd.DataFrame(data_c)
print(df_c.isnull().sum())
```

### Q11. Find the total number of missing values in the DataFrame.

```python
print(df_c.isnull().sum().sum())
```

### Q12. Replace missing Age values with the mean Age.

```python
df_c["Age"] = df_c["Age"].fillna(df_c["Age"].mean())
print(df_c)
```

### Q13. Replace missing Salary values with the median Salary.

```python
df_c["Salary"] = df_c["Salary"].fillna(df_c["Salary"].median())
print(df_c)
```

### Q14. Remove rows where Department is missing and display the cleaned DataFrame.

```python
df_c = df_c.dropna(subset=["Department"])
print(df_c)
```

# SECTION D – GROUPING AND AGGREGATION

### Q15. Calculate the average salary for each department.

```python
data_d = {
    "Employee": ["Arun", "Priya", "Karthik", "Divya", "Rahul", "Meena"],
    "Department": ["IT", "HR", "IT", "HR", "Finance", "Finance"],
    "Salary": [35000, 30000, 45000, 40000, 50000, 55000],
    "Experience": [2, 3, 5, 4, 6, 7]
}

df_d = pd.DataFrame(data_d)
print(df_d.groupby("Department")["Salary"].mean())
```

### Q16. Find the maximum salary in each department.

```python
print(df_d.groupby("Department")["Salary"].max())
```

### Q17. Count the number of employees in each department.

```python
print(df_d.groupby("Department")["Employee"].count())
```

### Q18. Calculate the average experience for each department.

```python
print(df_d.groupby("Department")["Experience"].mean())
```

### Q19. Sort departments by average salary in descending order.

```python
print(df_d.groupby("Department")["Salary"].mean().sort_values(ascending=False))
```

# SECTION E – ADVANCED PANDAS OPERATIONS

### Q20. Create a new column named Annual_Salary by multiplying monthly Salary by 12.

```python
df_d["Annual_Salary"] = df_d["Salary"] * 12
print(df_d)
```

### Q21. Rename the Employee column as Employee_Name.

```python
df_d = df_d.rename(columns={"Employee": "Employee_Name"})
print(df_d)
```

### Q22. Sort all employee records by Salary in descending order.

```python
df_d = df_d.sort_values(by="Salary", ascending=False)
print(df_d)
```

### Q23. Find the second-highest distinct salary using Pandas operations.

```python
second_highest_salary = df_d["Salary"].drop_duplicates().nlargest(2).iloc[-1]
print(second_highest_salary)
```
