16. Create a simple calculator that takes two numbers and an operator (+, -, *, /) and performs the operation.



```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
op = input("Enter operator (+, -, *, /): ")

if op == "+":
    print("Result:", a + b)
elif op == "-":
    print("Result:", a - b)
elif op == "*":
    print("Result:", a * b)
elif op == "/":
    if b != 0:
        print("Result:", a / b)
    else:
        print("Cannot divide by zero")
else:
    print("Invalid operator")
```

17. Get three subject marks and calculate the average. Print Pass only if the student scores at least 35 in every subject.



```python
mark1 = float(input("Enter mark for subject 1: "))
mark2 = float(input("Enter mark for subject 2: "))
mark3 = float(input("Enter mark for subject 3: "))

average = (mark1 + mark2 + mark3) / 3
print("Average:", average)

if mark1 >= 35 and mark2 >= 35 and mark3 >= 35:
    print("Pass")
else:
    print("Fail")
```

23. Get the electricity units consumed and calculate the bill using different unit slabs.



```python
units = float(input("Enter electricity units consumed: "))

if units <= 100:
    bill = units * 1.50
elif units <= 200:
    bill = 100 * 1.50 + (units - 100) * 2.50
elif units <= 300:
    bill = 100 * 1.50 + 100 * 2.50 + (units - 200) * 4.00
else:
    bill = 100 * 1.50 + 100 * 2.50 + 100 * 4.00 + (units - 300) * 6.00

print("Electricity bill: ₹", bill)
```

24. Get a student's marks and print the grade + pass/fail status.

25. Get the purchase amount and apply discount:
- Below ₹1,000 → No discount
- ₹1,000–₹4,999 → 10%
- ₹5,000–₹9,999 → 20%
- ₹10,000+ → 30%



```python
amount = float(input("Enter purchase amount: "))

if amount < 1000:
    discount_percent = 0
elif amount < 5000:
    discount_percent = 10
elif amount < 10000:
    discount_percent = 20
else:
    discount_percent = 30

discount = amount * discount_percent / 100
final_amount = amount - discount

print("Discount:", discount_percent, "%")
print("Discount amount: ₹", discount)
print("Final amount: ₹", final_amount)
```

26. Get a person's salary and calculate tax based on different salary slabs.



```python
salary = float(input("Enter salary: "))

if salary <= 250000:
    tax = 0
elif salary <= 500000:
    tax = (salary - 250000) * 0.05
elif salary <= 1000000:
    tax = 250000 * 0.05 + (salary - 500000) * 0.20
else:
    tax = 250000 * 0.05 + 500000 * 0.20 + (salary - 1000000) * 0.30

print("Tax: ₹", tax)
```

30. Create a movie ticket booking program:
- Ask for age
- Ask for number of tickets
- Calculate ticket price based on age category
- Apply a discount if more than 5 tickets are booked
- Print the final bill.



```python
age = int(input("Enter age: "))
tickets = int(input("Enter number of tickets: "))

if age < 5:
    price = 0
elif age <= 12:
    price = 50
elif age <= 59:
    price = 100
else:
    price = 70

total = price * tickets

if tickets > 5:
    discount = total * 0.10
else:
    discount = 0

final_bill = total - discount

print("Ticket price each: ₹", price)
print("Total: ₹", total)
print("Discount: ₹", discount)
print("Final bill: ₹", final_bill)
```
